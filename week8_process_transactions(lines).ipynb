```python
# Week 8: Smart Personal Finance Manager

def load_transactions(filename):
    """Read lines from a transaction file safely."""
    try:
        with open(filename, 'r') as file:
            lines = file.readlines()
        return [line.strip() for line in lines if line.strip()]
    except FileNotFoundError:
        print(f"File '{filename}' not found!")
        return []


def process_transactions(lines):
    """
    Convert raw lines into clean transaction dictionaries.
    Skips bad lines and counts errors.
    """
    transactions = []
    failed_lines = 0

    for line in lines:
        try:
            # Split by comma
            parts = line.split(",")

            # Check if valid format
            if len(parts) != 2:
                raise ValueError("Invalid format")

            description = parts[0]
            amount = float(parts[1])  # convert to number

            # Create dictionary
            transaction = {
                "description": description,
                "amount": amount
            }

            transactions.append(transaction)

        except:
            # If anything goes wrong, count it
            failed_lines += 1

    print(f"⚠️ Skipped {failed_lines} bad line(s)")
    return transactions


def show_summary(transactions):
    """Display summary of transactions."""
    
    total_income = 0
    total_expenses = 0

    for t in transactions:
        if t["amount"] > 0:
            total_income += t["amount"]
        else:
            total_expenses += t["amount"]

    net = total_income + total_expenses

    print("\n📊 Summary")
    print("=" * 30)
    print(f"Total Transactions: {len(transactions)}")
    print(f"Total Income: ${total_income:.2f}")
    print(f"Total Expenses: ${total_expenses:.2f}")
    print(f"Net Amount: ${net:.2f}")


# Main program
def main():
    print("🏦 Personal Finance Manager")
    print("=" * 30)

    filename = "transactions.txt"

    # Load data
    raw_lines = load_transactions(filename)

    if not raw_lines:
        print("No data to process!")
        return

    # Process data
    transactions = process_transactions(raw_lines)

    # Show results
    show_summary(transactions)


# Run program
main()
```
