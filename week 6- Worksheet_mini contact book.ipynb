```python
# Correct contacts dictionary
contacts = {
    "roman": {
        "name": "Roman",
        "phones": {
            "mobile": "555-1234"
        },
        "email": "roman.giri@gmail.com"
    },
    "bob": {
        "name": "Jenish",
        "phones": {
            "mobile": "555-5678"
        },
        "email": "jenish@gmail.com"
    },
    "charlie": {
        "name": "Charlie",
        "phones": {
            "mobile": "555-9012"
        },
        "email": "charlie@gmail.com"
    }
}

# Print all contacts correctly
for contact in contacts.values():
    print("Name:", contact["name"])
    print("Email:", contact["email"])
    print("Phones:", contact["phones"])
    print()
    search_name = input("Enter name to search: ")

if search_name in contacts:
    print("Phone:", contacts[search_name]["phone"])
    print("Email:", contacts[search_name]["email"])
else:
    print("Contact not found")
```
