```python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(celsius):

    # Apply conversion formula
    fahrenheit = (celsius * 9/5) + 32

    # Return result
    return fahrenheit


# Test cases
print(celsius_to_fahrenheit(0))      # 32.0
print(celsius_to_fahrenheit(25))     # 77.0
print(celsius_to_fahrenheit(100))    # 212.0


# Assertions to verify correctness
assert celsius_to_fahrenheit(0) == 32.0
assert celsius_to_fahrenheit(-40) == -40.0

print("All tests passed!")
```

```python
def count_vowels(text):

    vowels = "aeiou"
    count = 0

    for letter in text:

        print("Checking:", letter)

        if letter in vowels:
            print(letter, "is a vowel")
            count += 1

        print("Current count:", count)

    return count


print(count_vowels("Hello"))
```

```python
import re

def format_phone(phone):

    # Remove all non-digit characters
    digits = re.sub(r"\D", "", phone)

    # Validate phone number length
    if len(digits) != 10:
        return "Error: Invalid phone number"

    # Return formatted phone number
    return f"({digits[:3]}) {digits[3:6]}-{digits[6:]}"


# Print outputs
print(format_phone("1234567890"))
print(format_phone("123-456-7890"))
print(format_phone("123 456 7890"))
print(format_phone("(123) 456-7890"))
print(format_phone("12345"))
print(format_phone("abc1234567"))
print(format_phone(""))


# Assertions for testing
assert format_phone("1234567890") == "(123) 456-7890"
assert format_phone("12345") == "Error: Invalid phone number"

print("All test cases passed!")
```
